```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib

from sklearn.preprocessing import StandardScaler
from sklearn.feature_selection import RFE
```

```python
df = pd.read_csv('../data/processed/feature_engineered_data.csv')
```

```python
df['Date'] = pd.to_datetime(df['Date'])
```

```python
df
```

```python
df.info()
```

```python
train_size = int(0.8 * len(df))

train_df = df.iloc[:train_size].copy()
test_df  = df.iloc[train_size:].copy()
```

```python
date_train = train_df['Date'].reset_index(drop = True)
date_test  = test_df['Date'].reset_index(drop = True)
```

```python
feature_cols = [c for c in df.columns if c not in ['Date', 'Target']]
```

```python
x_scaler = StandardScaler()
y_scaler = StandardScaler()
```

```python
train_df[feature_cols] = x_scaler.fit_transform(train_df[feature_cols])
test_df[feature_cols]  = x_scaler.transform(test_df[feature_cols])

train_df['Target'] = y_scaler.fit_transform(train_df[['Target']]).ravel()
test_df['Target']  = y_scaler.transform(test_df[['Target']]).ravel()
```

```python
train_df.to_csv('../data/processed/train_scaled.csv', index = False)
test_df.to_csv('../data/processed/test_scaled.csv', index = False)

joblib.dump(x_scaler, '../models/x_scaler.pkl')
joblib.dump(y_scaler, '../models/y_scaler.pkl')

print(f"Train: {train_df.shape} | Test: {test_df.shape}")
```
