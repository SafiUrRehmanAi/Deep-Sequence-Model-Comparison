```python
import numpy as np
import pandas as pd
```

```python
train_df = pd.read_csv('../data/processed/train_scaled.csv')
test_df  = pd.read_csv('../data/processed/test_scaled.csv')
```

```python
feature_cols = [c for c in train_df.columns if c not in ['Date', 'Target']]
```

```python
SEQ_LEN = 10
```

```python
def make_sequences(df, seq_len):
    X = df[feature_cols].values
    y = df['Target'].values
    Xs, ys = [], []
    for i in range(len(df) - seq_len):
        Xs.append(X[i:(i + seq_len)])
        ys.append(y[i + seq_len])
    return np.array(Xs), np.array(ys)
```

```python
X_train, y_train = make_sequences(train_df, SEQ_LEN)
X_test,  y_test  = make_sequences(test_df, SEQ_LEN)
```

```python
np.savez(
    '../data/processed/lstm_ready.npz',
    X_train = X_train,
    y_train = y_train,
    X_test = X_test,
    y_test = y_test,
    feature_names = feature_cols,
    seq_len = SEQ_LEN
)

print(f"Train: {X_train.shape} | Test: {X_test.shape}")
```
